```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
data = {
    "CustomerID": [
        "C001","C002","C003","C004","C005",
        "C006","C007","C008","C009","C010",
        "C011","C012","C013","C014","C015",
        "C016","C017","C018","C019","C020"
    ],

    "Age": [
        22,45,31,52,27,
        39,24,48,35,29,
        55,26,42,33,51,
        23,37,46,30,58
    ],

    "AnnualIncome": [
        28000,85000,52000,92000,35000,
        68000,31000,105000,59000,45000,
        115000,33000,78000,61000,98000,
        27000,72000,88000,49000,120000
    ],

    "PurchaseFrequency": [
        8,22,15,25,10,
        19,7,28,17,13,
        30,9,21,16,26,
        6,20,24,14,32
    ],

    "TotalSpend": [
        12000,78000,42000,95000,18000,
        62000,10000,110000,50000,30000,
        135000,15000,72000,46000,88000,
        9000,65000,92000,38000,145000
    ]
}
df = pd.DataFrame(data)
print("=" * 60)
print("CUSTOMER DATA")
print("=" * 60)
display(df)
print("\nDATASET INFORMATION")
print("=" * 60)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("\nColumn Information:")
print(df.dtypes)
print("\nMISSING VALUES")
print("=" * 60)
print(df.isnull().sum())
df = df.drop_duplicates()
print("\nDuplicates removed.")
print("Current number of records:", len(df))
print("\nSTATISTICAL SUMMARY")
print("=" * 60)
display(df.describe())
features = [
    "Age",
    "AnnualIncome",
    "PurchaseFrequency",
    "TotalSpend"
]
X = df[features]
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
print("\nData standardization completed.")
inertia = []
cluster_range = range(2, 8)
for k in cluster_range:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)

    inertia.append(model.inertia_)


plt.figure(figsize=(9, 5))

plt.plot(
    cluster_range,
    inertia,
    marker="o"
)

plt.title(
    "Elbow Method for Optimal Number of Clusters"
)

plt.xlabel(
    "Number of Clusters"
)

plt.ylabel(
    "Inertia"
)

plt.grid(True)

plt.show()


# ============================================================
# 11. APPLY K-MEANS CLUSTERING
# ============================================================

# Four customer groups

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(
    X_scaled
)


silhouette = silhouette_score(
    X_scaled,
    df["Cluster"]
)

print("\nSILHOUETTE SCORE")
print("=" * 60)

print(
    round(silhouette, 3)
)

cluster_summary = df.groupby(
    "Cluster"
)[
    [
        "Age",
        "AnnualIncome",
        "PurchaseFrequency",
        "TotalSpend"
    ]
].mean()


print("\nCLUSTER SUMMARY")
print("=" * 60)

display(
    cluster_summary.round(2)
)

average_income = df["AnnualIncome"].mean()
average_spend = df["TotalSpend"].mean()
average_frequency = df["PurchaseFrequency"].mean()


segment_names = {}


for cluster in cluster_summary.index:

    income = cluster_summary.loc[
        cluster,
        "AnnualIncome"
    ]

    spend = cluster_summary.loc[
        cluster,
        "TotalSpend"
    ]

    frequency = cluster_summary.loc[
        cluster,
        "PurchaseFrequency"
    ]


    if (
        income >= average_income
        and
        spend >= average_spend
    ):

        segment_names[
            cluster
        ] = "High Value Customers"


    elif frequency >= average_frequency:

        segment_names[
            cluster
        ] = "Frequent Buyers"


    elif income < average_income:

        segment_names[
            cluster
        ] = "Budget Customers"


    else:

        segment_names[
            cluster
        ] = "Regular Customers"


df["Segment"] = df["Cluster"].map(
    segment_names
)

print("\nCUSTOMER SEGMENTS")
print("=" * 60)

display(
    df[
        [
            "CustomerID",
            "Age",
            "AnnualIncome",
            "PurchaseFrequency",
            "TotalSpend",
            "Cluster",
            "Segment"
        ]
    ]
)


segment_count = df[
    "Segment"
].value_counts()


print("\nCUSTOMER COUNT BY SEGMENT")
print("=" * 60)

display(
    segment_count
)

plt.figure(figsize=(10, 6))


for segment in df["Segment"].unique():

    segment_data = df[
        df["Segment"] == segment
    ]

    plt.scatter(
        segment_data["AnnualIncome"],
        segment_data["TotalSpend"],
        s=120,
        label=segment
    )


plt.title(
    "Customer Segmentation"
)

plt.xlabel(
    "Annual Income"
)

plt.ylabel(
    "Total Spend"
)

plt.legend()

plt.grid(True)

plt.show()

plt.figure(figsize=(10, 6))


plt.scatter(
    df["PurchaseFrequency"],
    df["TotalSpend"],
    s=120
)


plt.title(
    "Purchase Frequency vs Total Spend"
)

plt.xlabel(
    "Purchase Frequency"
)

plt.ylabel(
    "Total Spend"
)

plt.grid(True)

plt.show()

plt.figure(figsize=(9, 5))


plt.hist(
    df["Age"],
    bins=6,
    edgecolor="black"
)


plt.title(
    "Customer Age Distribution"
)

plt.xlabel(
    "Age"
)

plt.ylabel(
    "Number of Customers"
)

plt.grid(True)

plt.show()

segment_income = df.groupby(
    "Segment"
)["AnnualIncome"].mean()


plt.figure(figsize=(10, 5))


segment_income.plot(
    kind="bar"
)


plt.title(
    "Average Annual Income by Customer Segment"
)

plt.xlabel(
    "Customer Segment"
)

plt.ylabel(
    "Average Annual Income"
)

plt.xticks(
    rotation=20
)

plt.grid(
    axis="y"
)

plt.show()


segment_analysis = df.groupby(
    "Segment"
)[
    [
        "Age",
        "AnnualIncome",
        "PurchaseFrequency",
        "TotalSpend"
    ]
].mean().round(2)


print("\nSEGMENT CHARACTERISTICS")
print("=" * 60)

display(
    segment_analysis
)

print("\n")
print("=" * 60)
print("BUSINESS INSIGHTS")
print("=" * 60)


for segment in segment_analysis.index:

    row = segment_analysis.loc[
        segment
    ]


    print("\nSegment:", segment)

    print(
        "Average Age:",
        round(row["Age"], 1)
    )

    print(
        "Average Income: ₹",
        f"{row['AnnualIncome']:,.0f}"
    )

    print(
        "Purchase Frequency:",
        round(
            row["PurchaseFrequency"],
            1
        )
    )

    print(
        "Average Spending: ₹",
        f"{row['TotalSpend']:,.0f}"
    )


# ============================================================
# 23. OVERALL BUSINESS SUMMARY
# ============================================================

highest_spending_segment = (
    segment_analysis[
        "TotalSpend"
    ].idxmax()
)


highest_frequency_segment = (
    segment_analysis[
        "PurchaseFrequency"
    ].idxmax()
)


highest_income_segment = (
    segment_analysis[
        "AnnualIncome"
    ].idxmax()
)


print("\n")
print("=" * 60)
print("KEY FINDINGS")
print("=" * 60)


print(
    "\nHighest spending segment:",
    highest_spending_segment
)


print(
    "Most frequent purchasing segment:",
    highest_frequency_segment
)


print(
    "Highest income segment:",
    highest_income_segment
)


print(
    "\nCustomer segmentation analysis completed successfully."
)


df.to_csv(
    "customer_segmentation_results.csv",
    index=False
)


print(
    "\nResult saved as:"
)

print(
    "customer_segmentation_results.csv"
)
```
